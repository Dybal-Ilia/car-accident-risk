```python
import os
import sys
cwd = os.getcwd()
src_dir = os.path.join(cwd, '..', '../src/')
sys.path.append(src_dir)
from dataloader.data_load import download_dataset, map_data

download_dataset('playground-series-s5e10', '../data', unzip=True)
mapping = map_data('../data')
```

```python
data = mapping
```

# 1. Report Rendering

```python
from dqc.dqc import DQCLayer
dqc = DQCLayer()
validation_report = dqc.validation_report(data)
statistics_report = dqc.statistics_report(data)
report = dqc.render_report(validation_report=validation_report,
                          statistics_report=statistics_report)
```
